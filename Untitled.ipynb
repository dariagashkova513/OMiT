```python
# Run the notebook from the repository root.
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np

from tests.main import (
    DEFAULT_REGISTRY,
    create_model_runner,
    load_dataset_registries,
    load_registries,
)
from seg_eval.annotations import load_dataset
from seg_eval.metrics import box_iou


# -------------------------------------------------------------------
# Load the same model and dataset configuration used by tests.main
# -------------------------------------------------------------------
models = load_registries([DEFAULT_REGISTRY])
datasets = load_dataset_registries([Path("dataset_registry.yaml")])

dataset = load_dataset(datasets["bricks-coco"])

print("COCO classes:", list(enumerate(dataset.class_names)))

# Pick the first annotated image. Change this index to inspect other images.
sample_index = next(
    i for i, sample in enumerate(dataset.samples) if sample.instances
)
sample = dataset.samples[sample_index]

image_bgr = cv2.imread(sample.image_path)
if image_bgr is None:
    raise FileNotFoundError(sample.image_path)

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
height, width = image_bgr.shape[:2]

print("Image:", sample.image_path)
print("Image shape:", image_bgr.shape)
print("Ground-truth objects:", len(sample.instances))
print("Ground-truth class IDs:",
      sorted({instance.class_id for instance in sample.instances}))


# -------------------------------------------------------------------
# Exact automatic-grid setup from your command.
#
# The score threshold is lowered only for debugging so weak masks remain
# visible. Change it back to 0.88 afterward to reproduce filtering.
# -------------------------------------------------------------------
points_per_side = 4

runner = create_model_runner(
    "sam-vit-b",
    models,
    runtime_options={
        "points_per_side": points_per_side,
        "pred_iou_thres": 0.0,  # diagnostic; registry normally uses 0.88
        "min_mask_area": 1,     # diagnostic; registry normally uses 100
    },
)

result = runner(image_bgr)

print("SAM predictions retained:", len(result.instances))
print("Prediction class IDs:",
      sorted({instance.class_id for instance in result.instances}))
print("Prediction scores:",
      np.round([instance.score for instance in result.instances], 3))


# Recreate the 4x4 prompt locations used by tests/adapters/sam.py
xs = (np.arange(points_per_side) + 0.5) * width / points_per_side
ys = (np.arange(points_per_side) + 0.5) * height / points_per_side
grid_points = np.stack(np.meshgrid(xs, ys), axis=-1).reshape(-1, 2)


# -------------------------------------------------------------------
# Draw:
#   green boxes   = COCO ground truth
#   cyan crosses  = SAM point prompts
#   colored masks = SAM predictions
#   magenta boxes = SAM prediction boxes
# -------------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(20, 10))

axes[0].imshow(image_rgb)
axes[0].scatter(
    grid_points[:, 0],
    grid_points[:, 1],
    c="cyan",
    marker="x",
    s=80,
    linewidths=2,
    label="SAM prompts",
)

for gt in sample.instances:
    x1, y1, x2, y2 = gt.bbox
    rect = plt.Rectangle(
        (x1, y1),
        x2 - x1,
        y2 - y1,
        fill=False,
        edgecolor="lime",
        linewidth=1.5,
    )
    axes[0].add_patch(rect)

axes[0].set_title(
    f"Input: {len(sample.instances)} GT boxes and "
    f"{points_per_side**2} point prompts"
)
axes[0].legend()
axes[0].axis("off")


overlay = image_rgb.astype(np.float32).copy()
rng = np.random.default_rng(7)

for instance in result.instances:
    mask = instance.mask.astype(bool)
    color = rng.integers(40, 256, size=3)
    overlay[mask] = 0.45 * overlay[mask] + 0.55 * color

overlay = np.clip(overlay, 0, 255).astype(np.uint8)
axes[1].imshow(overlay)

for instance in result.instances:
    x1, y1, x2, y2 = instance.bbox
    rect = plt.Rectangle(
        (x1, y1),
        x2 - x1,
        y2 - y1,
        fill=False,
        edgecolor="magenta",
        linewidth=2,
    )
    axes[1].add_patch(rect)
    axes[1].text(
        x1,
        y1,
        f"{instance.score:.2f}",
        color="white",
        fontsize=9,
        bbox={"facecolor": "magenta", "alpha": 0.65, "pad": 1},
    )

axes[1].set_title(f"SAM masks retained: {len(result.instances)}")
axes[1].axis("off")

plt.tight_layout()
plt.show()


# -------------------------------------------------------------------
# Ignore classes temporarily and test whether prediction geometry overlaps
# any ground-truth brick.
# -------------------------------------------------------------------
if result.instances:
    best_ious = [
        max(
            (box_iou(gt.bbox, pred.bbox) for gt in sample.instances),
            default=0.0,
        )
        for pred in result.instances
    ]

    print("Best class-agnostic IoU for each prediction:")
    print(np.round(best_ious, 3))
    print("Predictions with IoU >= 0.50:", sum(x >= 0.50 for x in best_ious))
else:
    print(
        "No masks survived even with diagnostic thresholds. "
        "Inspect the encoder/decoder export and coordinate preprocessing."
    )
```
